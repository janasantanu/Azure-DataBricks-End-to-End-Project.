```python
p_file_name=dbutils.widgets.get("file_name")
```

### **Data Reading**

```python
df=spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format",'parquet')\
        .option("cloudFiles.schemaLocation",f'abfss://bronze@projecte2e.dfs.core.windows.net/checkpoint_{p_file_name}')\
        .load(f"abfss://source@projecte2e.dfs.core.windows.net/{p_file_name}")
```

## **Data writing**

```python
df.writeStream.format("parquet")\
    .outputMode('append')\
        .option("checkpointLocation",f"abfss://bronze@projecte2e.dfs.core.windows.net/checkpoint_{p_file_name}")\
            .option("path",f"abfss://bronze@projecte2e.dfs.core.windows.net/{p_file_name}")\
                .trigger(once=True)\
                    .start()
```
