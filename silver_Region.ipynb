# **Data reading from databricks schema**

```python
df=spark.read.table("project_cata.bronze.regions")
df.display()
```

```python
df=df.drop("_rescued_data")
df.display()
```

```python
df=df.split
```

# **Data writing**

```python
df.write.mode("overwrite")\
    .format("delta")\
        .save("abfss://silver@projecte2e.dfs.core.windows.net/regions")
```

```python
df=spark.read.format("delta")\
    .load('abfss://silver@projecte2e.dfs.core.windows.net/regions')
df.display()
```

```python
%sql
create table if not exists project_cata.silver.regions_silver
using delta
location "abfss://silver@projecte2e.dfs.core.windows.net/regions"
```
